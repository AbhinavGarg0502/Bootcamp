```python
# String and Word Reversal

# 1. Create a string
original_text = "Python Programming is Fun"
print(f"Original String:               '{original_text}'")

# 2. Print in reverse order (character-by-character)
reversed_text = original_text[::-1]
print(f"Reversed String:               '{reversed_text}'")

# 3. Reverse the order of words
words_reversed_order = " ".join(original_text.split()[::-1])
print(f"Reversed Word Order:           '{words_reversed_order}'")

# 4. Reverse each individual word
each_word_reversed = " ".join(word[::-1] for word in original_text.split())
print(f"Each Individual Word Reversed: '{each_word_reversed}'")
```

```python
# List of random numbers and finding even numbers
import random

# 1. Generate a list of 10 random numbers between 1 and 100
random_numbers = [random.randint(1, 100) for _ in range(10)]
print(f"List of Random Numbers: {random_numbers}")

# 2. Extract and print all even numbers
even_numbers = [num for num in random_numbers if num % 2 == 0]
print(f"Even Numbers:           {even_numbers}")
```
